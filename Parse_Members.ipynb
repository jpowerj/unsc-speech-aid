```python
import pandas as pd
```

```python
member_df = pd.read_csv('membership/DPPA-SCMembership.csv')
```

```python
member_df = member_df.drop(columns='last_update')
```

```python
p5 = [
  'United States of America',
  'France',
  'China',
  'United Kingdom of Great Britain and Northern Ireland',
  'Russian Federation',
  'USSR (Union of Soviet Socialist Republics)',
]
```

```python
em_df = member_df[~member_df['security_council_member'].isin(p5)].copy()
```

```python
# Good lord... we have to manually specify the NaNs
id_map = {
  202: "Côte d'Ivoire",
  212: "Côte d'Ivoire",
  475: "Democratic Republic of the Congo",
  489: "Democratic Republic of the Congo",
  584: "Côte d'Ivoire",
  595: 'Democratic Republic of the Congo',
  599: "Côte d'Ivoire",
  609: 'Democratic Republic of the Congo',
  1003: "Côte d'Ivoire",
  1018: "Côte d'Ivoire",
}
def name_from_id(row):
  if row['id'] in id_map:
    return id_map[row['id']]
  return row['security_council_member']
em_df['security_council_member'] = em_df.apply(name_from_id, axis=1)
```

```python
em_df[pd.isna(em_df['security_council_member'])]
```

```python
em_df = em_df.rename(columns={'security_council_member': 'country'})
```

```python
name_map = {
  'Bolivia (Plurinational State of)': 'Bolivia',
  'Venezuela (Bolivarian Republic of)': 'Venezuela',
  'Syrian Arab Republic': 'Syria',
  'Ukrainian SSR (Ukrainian Soviet Socialist Republic)': 'Ukraine',
  'United Republic of Tanzania': 'Tanzania',
  'German FR (German Federal Republic)': 'West Germany',
  'Iran (Islamic Republic of)': 'Iran',
  'German DR (German Democratic Republic)': 'East Germany',
  'United Arab Republic': 'Egypt', #sigh
  'Republic of Korea': 'South Korea',
  'Congo': 'Republic of the Congo',
}
em_df['country'] = em_df['country'].apply(lambda x: name_map[x] if x in name_map else x)
```

```python
em_df.to_csv('elected_members.csv', index=False)
```

```python
em20_df = em_df[em_df['year'] <= 2020].copy()
```

```python
em20_df.to_csv('elected_members_2020.csv', index=False)
```
