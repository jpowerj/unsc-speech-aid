## Meetings

```python
meeting_df = pd.read_csv('PRIO_files/meetings.tsv', delimiter='\t', encoding='utf-8')
```

```python
meeting_df.columns
```

```python
meeting_df = meeting_df.drop(columns=['Unnamed: 0'])
```

```python
meeting_df.head(2)
```

```python
meeting_df = meeting_df.rename(columns={'word_count': 'total_word_count'})
```

```python
meeting_df.to_parquet('unsc_meetings.parquet', index=False)
```

## Big big merge

```python
merged_df = unsc_df.merge(
  meeting_df,
  on='record_id',
  how='left',
  suffixes=('','_meeting'),
  indicator=True,
)
```

```python
merged_df['_merge'].value_counts()
```

```python
merged_df.head()
```

```python
merged_df.to_parquet('unsc.parquet', index=False)
```
